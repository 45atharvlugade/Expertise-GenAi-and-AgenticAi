Set Operations.

1- Union.
2- Intersection.
3- Difference.
4- Syymetric Difference.

```python
a={1,2,3,4,5}
b={4,5,6,7,8}
```

1- Union.

```python
ans1=a|b
print(ans1,type(ans1))

# if common elements are taken only once.
```

Intersection

```python
ans2=a&b

print(ans2)

# returns only the common elements fromt he both sets.
```

Difference

```python

ans3=a-b
print(ans3)

# removes the common elements from it.
```
