Dictionary
 Dictionary is a mutable data type in Python.
 A python dictionary is a collection of key and value pairs separated by a colon (:) & enclosed
 in curly braces {}.
 Keys must be unique in a dictionary, duplicate values are allowed.

Sytax to create the Dict ,
1- using the constructor and 
2- normal creation.

```python
# using the constructor.
d1=dict()

# normal creating syntax 
d2={}


print(type(d1),type(d2))
```

```python
d1 = {1:'one' , 2:'two' , 3:'three'}
```

```python
print(d1)
```

```python
# to print the Keys from the Dict.
print(d1.keys())
```

```python
# to print the values from the keys.
print(d1.values())
```

```python
# can we store the list tupe or set as the values of the keys on the dict ?
d2={1:[1,2,3,4,5],2:(6,7,8,9,0),3:{6,54,73},4:d1}
```

```python
print(d2.keys(),':',d2.values())
```

Now Lets Access the dict elements

```python
# accessing the items using the keys.
print(d1[1])

# accessing the items usng the get method.
print(d1.get(1))
```
