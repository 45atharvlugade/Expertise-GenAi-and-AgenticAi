SET OPERATIONS ON SET.

```python
s1={4,3,101,100,2,5,6,7,8,9}
s2={5,3,7,9,1,4,7,10}
s3={13,15,54,14}
print(s1,s2,s3)
```

```python
# union
s4=s2.union(s1)
print(s4)
s5 = s1.union(s2)
print(s5)

print(s1|s2) # it is alos union
```

```python
# intersection 

print(s1.intersection(s2))
print(s1 & s2)   # intersection symbol -> print all the command values present in bith the sets
```

```python
# difference
print(s1.difference(s2))
print(s1-s2)
```

```python
# symmetric difference  : which ever the commons are removed and others are printed from both

print(s2.symmetric_difference(s3))
```

```python
print(len(s3))
```

```python
# symmetric difference update

s1.symmetric_difference_update(s3)
print(s1)
```

```python
# super set.
s1.issuperset(s2)

# subset
s1.issubset(s2)

s1.isdisjoint(s2)
```
