List is heterogenous data structure.
List allows duplicate elements.

```python
#List
list = [1,4,6,8]
print(list)
print(type(list))
```

```python
# how to get the length odf list
print(len(list))
```

```python
# adding the numbers to the list

list.append(5)
print(list)
```

```python
# remove the element from the list
# if multiples elements are present then it will 
# not remove all the elements from the list instead it will remmove the elements one by one 


list.remove(5)
print(list)
```

```python
# can we add the other data type in list

# yes we can add the other data type elements in it
list.append('str')
print(list)
```

```python
l1=[3,2,5,7]
l2= list.copy()
print(l1,l2)
```

```python
# clear funtion in List.

l3=[2,3,4,5,6,8]
print(l3)
l3.clear()
print(l3)
```

```python
# count function 

l4 = [1,2,1,2,3,4,5,5,6,6,4,2,7,8,5]
print(l4.count(3))
print(l4.count(1))
print(l4.count(5))
```

```python
# del keyword

del(l4)
print(l4)
```

```python
# SLICING IN LIST

l5=[1,2,3,4,5,6,7,8,9]
print(l5[:])# print all the elements 
print(l5[2:]) # print the elements from the 2 nd index
print()
```
