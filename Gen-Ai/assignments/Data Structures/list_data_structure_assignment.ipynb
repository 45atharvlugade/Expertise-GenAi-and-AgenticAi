🟢 Level 1: Basic Assignments
Assignment 1: Sum of Elements
- Create a list of integers.
- Calculate the sum without using sum().
Example: [10, 20, 30, 40, 50]
Output: 150

Assignment 2: Count Even and Odd Numbers
- Count how many even and odd numbers are present in a list.
Example: [1, 2, 3, 4, 5, 6, 8]
Output: Even: 4, Odd: 3

Assignment 3: Reverse a List
- Reverse the list without using reverse() or slicing.
Example: [10, 20, 30, 40, 50]
Output: [50, 40, 30, 20, 10]

Assignment 4: Find the Largest Element
- Find the largest number without using max().
Example: [12, 45, 7, 89, 23]
Output: 89

Assignment 5: Remove Duplicates
- Create a new list containing only unique elements.
- Maintain the original order.
Example: [1, 2, 3, 2, 4, 1, 5]
Output: [1, 2, 3, 4, 5]

Assignment 6: Search an Element
- Take a number from the user.
- Check whether it exists in the list without using the in operator.
Example: [10, 20, 30, 40, 50]
Search: 30
Output: Element found at index 2
🟡 Level 2: Intermediate Assignments
Assignment 7: Find the Second-Largest Element
- Find the second-largest distinct number without sorting.
Example: [10, 25, 8, 40, 25, 30]
Output: 30

Assignment 8: Move Zeros to the End
- Move all zeros to the end.
- Preserve the order of the other elements.
Example: [0, 1, 0, 3, 12, 0, 5]
Output: [1, 3, 12, 5, 0, 0, 0]

Assignment 9: Find Common Elements
- Find the elements present in both lists without repeating them in the result.
Example:
a = [1, 2, 3, 4, 5]b = [3, 4, 5, 6, 7]


Output: [3, 4, 5]

Assignment 10: Rotate a List
- Rotate the list to the right by k positions.
Example: [1, 2, 3, 4, 5], k = 2
Output: [4, 5, 1, 2, 3]

Assignment 11: Find the Frequency of Each Element
- Count how many times each number appears.
- Do not use count().
Example: [1, 2, 2, 3, 1, 2, 4]

Output:
1: 2 times
2: 3 times
3: 1 time
4: 1 time

A list contains distinct numbers from 1 to n, with exactly one number missing. Find the missing number.

Example: [1, 2, 4, 5, 6]

Example: [3, -1, 0, -5, 8, 0, -2, 4]
Output: [3, 8, 4, -1, -5, -2, 0, 0]
